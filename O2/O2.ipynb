```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from uncertainties import ufloat
from uncertainties.umath import sqrt
from odrpack import odr_fit
from scipy.optimize import curve_fit
from uncertainties import unumpy as unp
%matplotlib inline
plt.ion()
```

```python
def u_print(*args):
    print(*[f"{x:.2uL}" for x in args])
```

```python
obj = ufloat(23.6,0.1)
def f(s,o):
    socz = ufloat(s,0.1)
    obr=ufloat(o,0.1)
    a=socz-obj
    b=obr-obj
    return a*b/(a+b)
vf = np.vectorize(f)

soczewki = [40.0,39.6,38.8,38.7,38.4]
obrazy = [97.2,100.0,104.0,110.0,120.0]

result1 = vf(soczewki,obrazy)
#u_print(result1)

r_soczewki = [41.2,41.1,40.3,40.1,39.8]
r_obrazy = [97.2,100.0,104.0,110.0,120.0]

result2=vf(r_soczewki,r_obrazy)
#u_print(result2)

u_print(1/(np.mean(result1)/100))

u_print(np.mean(result1),np.mean(result2))
u_print((np.mean(result1)+np.mean(result2))/2)
```

```python
def Bessel(s_small, s_big,o):
    obr = ufloat(o,0.1)
    socz_s = ufloat(s_small,0.1)
    socz_b = ufloat(s_big,0.1)
    l=obr-obj
    return (l**2 - (socz_s-socz_b)**2)/(4*l)
vBessel = np.vectorize(Bessel)
```

```python
big = [41.2,40.9,40.3,40.0,39.6]
small = [80.4,83.1,87.6,93.4,104.5]
obraz  = [97.2,100.0,104.0,110.0,120.0]

result = vBessel(small,big,obraz)
f_1 = np.mean(result)
print(result)
u_print(f_1)
print(100/(result))
u_print(100/f_1)
```

```python
b_d = [47.7,49.2,50.5,53.0,46.8]
s_d = [95.9,83.7,77.3,72.5,107.0]
o_d = [120,110,104,100,130]

f_res = vBessel(s_d,b_d,o_d)
f_r = np.mean(f_res)
print(f_res,f_r)
u_print(f_r)

f_2= 1/(1/f_r - 1/f_1)
u_print(f_2)
u_print(100/f_2)
```
